#Q4  Build the Feature — Surge Pricing Function

```python
def calculate_fare(distance, time_of_day, is_raining, is_weekend):
  base_fare = 0
  if distance <=5:
    base_fare = 5
  elif distance<=15:
    base_fare = 10
  else:
    base_fare = 15

  if time_of_day == "night":
    base_fare+= 3

  if is_raining and distance> 10:
    base_fare+= base_fare*0.2

  if is_weekend:
    base_fare+= base_fare*0.15

  print(f"your base fare is: {base_fare:.2f}")
```

```python
calculate_fare(15,"night",True,False)
```

```python
calculate_fare(15,"day",True,False)
```

```python
calculate_fare(15,"day",True,True)
```

##Q6  Analyze & Build

```python
daily_followers = [500, 515, None, 540, 570, 565, 610]
total_growth = 0

for i in range(len(daily_followers)):

  if daily_followers[i] == None :
    daily_followers[i] = daily_followers[i-1]
    continue

  total_growth += daily_followers[i]

  if daily_followers[i] - daily_followers[i-1]  > 20 and i != 0:
    print("🚀 Viral Day detected!")

highest = daily_followers.index(max(daily_followers))

print(f"total: {total_growth}, highest: {highest}")
```

##Q7  Design the Classes

```python
class Animal:
  def __init__(self,name, species, age, hunger_level= 5):
    self.name = name
    self.species = species
    self.age = age
    self.hunger_level = hunger_level

  def feed(self, amount):
    self.hunger_level -= amount
    if self.hunger_level < 0 :
      self.hunger_level = 0
    print(f"hunger level was reduced to {self.hunger_level}")

  def describe(self):
    return f"{self.name} is a {self.age}-year-old {self.species} with hunger level {self.hunger_level}."
```

```python
class Lion(Animal):
  def roar(self):
    print(f"{self.name} lets out a mighty ROAR!")
  def describe(self):
    return f"{super().describe()} (King of the Jungle)"
```

```python
lion1 = Lion("simba","lion",4,7)
lion1.feed(4)
lion1.feed(5)
lion1.roar()
print(lion1.describe())
```
