## Problem 1: Separate Letters and Numbers

Ask the user for a string that contains both letters and numbers (e.g., "Py2th0on3"). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word.

**Example:**
- Input: "Py2th0on3"
- Output: Word: "Python", Sum: 5

````python
letter =[]
num=[]
text=input("enter a text ")
for char in text:
  if char.isalpha():
    letter.append(char)
  elif char.isnumeric():
    num.append(int(char))

print(f'Word: "{"".join(letter)}", Sum: {sum(num)}')
````

## Problem 2: Password Validation

Ask a user to enter a password. Check if it is valid based on these rules:
- Length must be between 8 and 16 characters
- It must contain at least one uppercase letter
- It must contain at least one special character (e.g., @, #, $)

**Example:**
- Input: "Guest@123"
- Output: "Password is Secure."

````python
password = input("enter a password: ")
if (
    8 <= len(password) <= 16
    and any(char.islower() for char in password)
    and any(not char.isalnum() for char in password) #including space as a special character
    ):
  print("Password is Secure.")
else:
  print("Password is not Secure.")
````

## Problem 4: Common Names in Lists

Ask the user to input two separate lists of names (comma-separated). Create a third list that contains only the names that appear in both lists.

**Example:**
- Input: List 1: "Ali, Ben, Sara", List 2: "Ben, Joy, Ali"
- Output: Common names: ['Ali', 'Ben']

````python
list1 = input("enter a list of names (comma-separated): ").split(",")
list2 = input("enter another list of names (comma-separated): ").split(",")
common_names = []

list1 = [(name.strip()).lower() for name in list1]
list2 = [(name.strip()).lower() for name in list2]

for name1 in list1:
  for name2 in list2:
    if name1 == name2 and name1 not in common_names:
      common_names.append(name1)

print(f"Common names: {common_names}")
````

## Problem 5: Student Test Averages

Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the average for each student. If the average is > 90, print the student's name followed by "Honor Roll."

**Example:**
- Input: {'Sam': [95, 90, 92]}
- Output: Sam: Average 92.3 - Honor Roll

````python
students = {
    "sam": [95,90,92],
    "mari": [85,89,90],
    "hajar":[98,100,99]
}

for student in students:
  grades = students[student]
  avg = sum(grades) / len(grades)
  honor = ""
  if avg > 90:
    honor = "Honor Roll"

  print(f"{student}: Average {avg:.1f} - {honor}")
````

## Problem 8: Element Frequency Counter

Given a list of numbers with many duplicates, return a list of tuples showing each unique element and its frequency (how many times it appeared).

**Example:**
- Input: [1, 1, 2, 3, 3, 3, 4]
- Output: [(1, 2), (2, 1), (3, 3), (4, 1)]

````python
nums = [1, 1, 2, 3, 3, 3, 4]
nums_set = set(nums)
counter = []

for num in nums_set:
  counter.append((num , nums.count(num)))

print(counter)
````

## Problem 9: Factorial and Sum Divisibility

Ask the user for a number $n$. Calculate the factorial (n!) using a loop, and then check if the resulting factorial is divisible by the sum of numbers from 1 to $n$.

**Example:**
- Input: 4
- Output: Factorial is 24, Sum is 10. 24 is not divisible by 10.

````python
n = int(input("enter a number: "))
sum =0
fact=1
while n>0:
  fact*= n
  sum+= n
  n-=1

if fact%sum == 0 :
  print(f"Factorial is {fact}, Sum is {sum}. {fact} is divisible by {sum}.")
else:
  print(f"Factorial is {fact}, Sum is {sum}. {fact} is not divisible by {sum}.")
````

## Problem 10: Shopping Cart System

Start with a dictionary of items and prices: {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}. Allow the user to "Add" or "Remove" items from a cart. If they add an item not in the store, print an error. Finally, print the total bill.

**Example:**
- Input: Add 'apple', Add 'milk'
- Output: Total Bill: $2.0

````python

items = {'apple': 0.5,
          'bread': 2.0,
          'milk': 1.5}
cart = []
sum = 0
while True:
  user_input = input("enter wether you want to add, remove, print total bill or exit: ")
  if user_input == "add":
    item = input("enter the item to add: ")
    if item in items :
      cart.append(item)
      sum += items[item]
      print(f"{item} was added successfully")
      print("Cart:",cart)
    else:
      print("enter a valid item")

  elif user_input == "remove":
    item = input("enter and item you want to remove: ")
    if item in cart:
      cart.remove(item)
      sum -= items[item]
      print(f"{item} was removed successfully")
      print("Cart:",cart)
    else:
      print(f"{item} does not exist in your cart")

  elif user_input == "total":
    print(f"Total Bill: ${sum}")

  elif user_input == "exit":
    print(f"Cart:",cart)
    print(f"Total Bill: ${sum}")
    print("BYE!")
    break
  else:
    print("invalid operator")

````

## Problem 15: Vowel Index Mapper

Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear.

**Example:**
- Input: "banana"
- Output: {'a': [1, 3, 5]}

````python
vowels = {
    'a':[],
    'e':[],
    'i':[],
    'o':[],
    'u':[]
}
vowels_count ={}

word = input("enter a word: ").lower()
for i in range(len(word)):
  if word[i] in vowels:
    vowels[word[i]].append(i)

for vowel in vowels:
  if  vowels[vowel]:
    vowels_count[vowel] = vowels[vowel]

print(vowels_count)
````
