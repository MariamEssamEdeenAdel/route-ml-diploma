1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.
   - **Input:** `list=["apple", "pie", "banana"], K=3`
   - **Output:** `['apple', 'banana']`

```python
def strlength(strings , k):
  new_strings = []
  for item in strings:
    if len(item) > k:
      new_strings.append(item)
  return new_strings
```

```python
strlength(["apple", "pie", "banana"],3)
```

2. Write a function named `area()` that calculates the area of either a "circle" or a "rectangle". The function should take the shape name as the first argument, followed by the necessary dimensions (radius for circle; length and width for rectangle).
   - **Example Input:** `area("circle", 5)` OR `area("rectangle", 4, 6)`
   - **Example Output:** `Circle Area: 78.5`, `Rectangle Area: 24`

```python
def area(shape , length , width=0):
  if shape == "circle":
    return f'Circle Area: {3.14 * length**2}'
  elif shape == "rectangle":
    return f'Rectangle Area: {length * width}'
```

4. Write a program to find the sum of all numbers in a list. To make it mid-level, ensure the program ignores any elements in the list that are not numbers (like strings or booleans).
   - **Example Input:** `[10, "apple", 20, True, 5]`
   - **Example Output:** `Total Sum: 35`

```python
def sum_num(list_num):
  total = 0
  for item in list_num:
    if str(item).isnumeric():
      total+=item
  return f"Total Sum: {total}"
```

5.	Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.)
Perform the following tasks now:
-	Now add items to the menu.
-	Make table reservations.
-	Take customer orders.
-	Print the menu.
-	Print table reservations.
-	Print customer orders.


```python
class Restaurant:
  def __init__(self):
    self.menu_items = {}
    self.booked_tables =[]
    self.customer_orders = {}

  def add_item_to_menu(self, name , price):

    self.menu_items[name] = price
    print(f"{name} was added successfully with price {price}$")


  def book_tables(self, table_num, num_people, order):
    self.customer_orders[len(self.customer_orders)+1] = order
    reservation = {"id": len(self.booked_tables)+1 ,
                   "order": order ,
                   'table_num': table_num,
                   "number_of_people" : num_people,
                   'order_id' : len(self.customer_orders)+1
                   }
    self.booked_tables.append(reservation)

    print(f"table num {table_num} was successfully booked for {num_people} people\n with order {order}")



  def customer_order(self,order):
    self.customer_orders[len(self.customer_orders)+1] = order
    print(f"order id {len(self.customer_orders)+1} order {order}")


  def display_menuItems(self):
    print("Menu")
    for key,value in self.menu_items.items():
      print(f"{key} : {value}$")


  def display_reservations(self):
    print("Reservations:")
    for item in self.booked_tables:
      print(f"""
      Reservation number {item['id']}
      Table number {item['table_num']}
      Number of people {item['number_of_people']}
      Order {item['order']}
            """)

  def display_orders(self):
    print("Orders")
    for key,value in self.customer_orders.items():
      print(f"""
      Order ID: {key}
      Order: {value}
      """)
```

```python
res = Restaurant()
res.add_item_to_menu("pasta", 14)
res.add_item_to_menu("pizza", 14)
res.add_item_to_menu("sushi", 14)
res.add_item_to_menu("salad", 14)
```

```python
res.display_menuItems()
```

```python
res.book_tables(1, 4, ['pasta','sushi','salad','pizza'])
```

```python
res.book_tables(2, 2, ['pasta','sushi'])
```

```python
res.display_reservations()
```

```python
res.customer_order(['salad','sushi'])
```
