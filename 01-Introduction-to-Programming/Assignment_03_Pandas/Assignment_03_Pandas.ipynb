## Task 2: Handling Missing Values

The dataset contains missing values in the `Salary` and `Bonus` columns.

**Requirements:**
- Fill the missing `Salary` values with the median salary of the entire company
- Fill the missing `Bonus` values with 0

```python
df['Salary'] =  df['Salary'].fillna(df['Salary'].mean())
```

```python
df.isna().sum()
```

```python
df['Bonus'] = df['Bonus'].fillna(0)
```

```python
df.isna().sum()
```

## Task 3: Data Cleaning

The `Employee_Name` column is in the format "FirstName LastName".

**Requirements:**
- Split this column into two new columns: `First_Name` and `Last_Name`
- Delete the original `Employee_Name` column

```python
df['Employee_Name'] = df['Employee_Name'].map(lambda x: x.split() )
df['Employee_Name']
```

```python
df['First_Name']= df['Employee_Name'].map(lambda x: x[0] )
df['Last_Name'] = df['Employee_Name'].map(lambda x: x[1] )
df.head()
```

```python
df = df.drop("Employee_Name" , axis=1)
df.head()
```
