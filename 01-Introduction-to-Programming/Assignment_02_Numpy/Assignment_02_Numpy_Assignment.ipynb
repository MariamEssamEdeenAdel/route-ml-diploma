## Problem 1: Array Normalization

Create a 3x3 array with random integers between 1 and 100. Write a program to "normalize" the array so that the values are scaled between 0 and 1.

**Formula:** `(x - min) / (max - min)`

**Example Output:** An array where the smallest value becomes 0.0 and the largest becomes 1.0.

````python
rand_int_array = np.random.randint(1,101 , size=(3,3))
minimum = rand_int_array.min()
maximum = rand_int_array.max()
normalized_array = (rand_int_array - minimum) / (maximum - minimum)

print(normalized_array)
````

## Problem 2: Find Closest Value

Create a 1D array of 10 random numbers. Ask the user for a value X. Find the value in the array that is closest to X.

**Example Input:**
- Array: `[1, 5, 8, 12]`
- X = 10

**Example Output:** 8 or 12 (whichever is mathematically closer)

````python
#I am assuming from the example  that numbers between 1 and 10 and integers
x= int(input("enter a number between 1 and 10: "))
array_ = np.random.randint(1,10,10)
print("array: ",array_)
array2= abs(x - array_)
closest_gap = array2.min()
closest_index = list(array2).index(closest_gap)
print("closest number: ", array_[closest_index])
````

## Problem 3: Second-Largest Value in Each Column

Write a NumPy program to create a 5x5 array with random values. For each column in the array, find the second-largest value.

**Example Input:**
```
[[1, 2, 3],
 [4, 5, 6],
 [7, 8, 9]]
```

**Example Output:** `[4, 5, 6]` (These are the second-highest values for each respective column)

````python
arr = np.random.randint(1,11,(5,5))
print("Array:\n", arr)

arr_sorted = np.sort(arr,axis=0)

second_highest= arr_sorted[-2]
print("second highest: " ,second_highest)
````

## Problem 4: Find Rows with Even Sum

Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows.

````python
matrix=np.random.randint( 1,11,size = (6,4))
print("matrix:\n",matrix)
rows_sum = matrix.sum(axis=1)
print("sums: " ,rows_sum)
is_even = np.array(rows_sum%2 == 0)

indices= np.where(is_even) #we have not learned about it but I searched about a numpy method that returns an index
print(indices)
````

## Problem 5: Replace Multiples

Create a 1D array of 30 random integers between 1 and 50. Replace all numbers that are multiples of 3 or 5 with the value -1.

**Example Input:** `[3, 7, 10, 15]`

**Example Output:** `[-1, 7, -1, -1]`

````python
arr = np.random.randint(1,51,size=30)
print("origin array: ",arr)
arr[arr%3==0 ] = -1
arr[arr%5==0] = -1
print("Replace Multiples array: ",arr)
````

## Problem 6: Row Mean Subtraction

Write a NumPy program to create a 3x3 array with random values. For each row, calculate its mean (average) and subtract that mean from every element in that specific row.

**Example Input:**
```
[[10, 20, 30],
 [5, 5, 5],
 [0, 10, 20]]
```

**Example Output:**
```
[[-10.  0. 10.],
 [ 0.  0.  0.],
 [-10.  0. 10.]]
```

````python
arr = np.random.randint(1,21,size=(3,3))
print("original array:\n",arr)
means = np.reshape(arr.mean(axis=1) , (3,1))
print("means array:\n",means)
print()
new_arr = arr - means
print("Row Mean Subtraction array:\n",new_arr)
````
